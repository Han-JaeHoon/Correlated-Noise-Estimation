```python
import pennylane as qml
import numpy as np
import random
```

```python
n_qubits = 2

dev = qml.device("default.clifford", wires=n_qubits, tableau=False)
```

```python
@qml.qnode(dev)
def test_circuit():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.probs(wires=[1])
```

```python
result = test_circuit()
print(result)
```

```python
qml.draw_mpl(test_circuit)()
```
